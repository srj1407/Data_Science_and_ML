```python
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X = scaler.fit_transform(df)
```

```python
scaled_df = pd.DataFrame(X, columns = df.columns, index = df.index)
```

```python
scaled_df.head()
```

```python
import scipy.cluster.hierarchy as sch

Z = sch.linkage(scaled_df, method = 'ward')
```

```python
Z.shape
```

```python
Z
```

```python
fig, ax = plt.subplots(figsize = (20, 12))
sch.dendrogram(Z, labels = scaled_df.index, ax = ax, color_threshold = 2)
plt.xticks(rotation = 90)
ax.set_ylabel('Distance')
```

```python
fig, ax = plt.subplots(figsize = (20, 12))
sch.dendrogram(Z, labels = scaled_df.index, ax = ax, color_threshold = 2)
plt.xticks(rotation = 90)
plt.axhline(y = 5, color = 'k', linestyle = '--')
ax.set_ylabel('Distance')
```

```python
from sklearn.cluster import AgglomerativeClustering

ac = AgglomerativeClustering(n_clusters = 5, metric = 'euclidean', linkage = 'ward')
y_pred = ac.fit_predict(scaled_df)
```

```python
viz_df = scaled_df
viz_df['y_pred'] = y_pred
```

```python
viz_df.head()
```

```python
clustered_df = viz_df.groupby('y_pred').mean()
```

```python
clustered_df
```

```python
clustered_df.T
```

```python
labels = ['Cluster 1', 'Cluster 2', 'Cluster 3', 'Cluster 4', 'Cluster 5']

plt.figure(figsize = (14, 8))
plt.plot(clustered_df.T, label = labels)
plt.xticks(rotation = 90)
plt.legend(labels)
```

```python
from sklearn.metrics import silhouette_score

def hierarchical_clustering_silhouette(scaled_df, n_clusters):
  clustering = AgglomerativeClustering(n_clusters = n_clusters)
  labels = clustering.fit_predict(scaled_df)
  silhouette_avg = silhouette_score(scaled_df, labels)
  return silhouette_avg

for n_clusters in range(2, 9):
  silhouette_avg = hierarchical_clustering_silhouette(scaled_df, n_clusters)
  print(f'n_clusters: {n_clusters} | Silhouette score: {silhouette_avg}')
```

```python
scaled_df.head()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components = 2)
pca_data = pca.fit_transform(scaled_df)
labels = scaled_df['y_pred']

pca_df = pd.DataFrame(data = np.vstack((pca_data.T, labels)).T, columns = ('1st_principal', '2nd_principal', 'label'))
```

```python
pca_df
```

```python
import seaborn as sns

g = sns.FacetGrid(pca_df, hue="label",height=5 ,aspect= 2)
g.map(plt.scatter, '1st_principal', '2nd_principal')
plt.show()
```

```python
from sklearn.manifold import TSNE

tsne = TSNE(n_components = 2, random_state = 0, perplexity = 50)
tsne_data = tsne.fit_transform(scaled_df)
```

```python
tsne_df = pd.DataFrame(data = np.hstack((tsne_data, scaled_df['y_pred'].values.reshape(-1, 1))), columns=("Dim_1", "Dim_2", "label"))
```

```python
tsne_df.head()
```

```python
# Plot using Seaborn
sns.FacetGrid(tsne_df, hue="label", height=5 ,aspect= 2).map(plt.scatter, 'Dim_1', 'Dim_2').add_legend()
plt.title('With perplexity = 50')
plt.show()
```

```python
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (16,12)
```

```python
df = pd.read_csv('./E-commerce.csv')
df.info()
```

```python
std_scaler = StandardScaler().fit(df)
std_df = std_scaler.transform(df)
std_df = pd.DataFrame(std_df,columns=df.columns)

std_df.head()
```

```python
start = time.time()
dbs = DBSCAN(eps = 1.0, min_samples = 12).fit(std_df)
end = time.time()
```

```python
labels = dbs.labels_
labels
```

```python
out_df = std_df
out_df['labels'] = labels
```

```python
out_df['labels'].value_counts()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components = 2)
X = pca.fit_transform(std_df)
```

```python
X
```

```python
x = X[:, 0]
y = X[:, 1]

fig, ax = plt.subplots()
sc = ax.scatter(x,y,s=40,c=out_df['labels'],cmap='viridis')
ax.legend(*sc.legend_elements(), title='clusters')
plt.show()
```

```python
from sklearn.manifold import TSNE

tsne = TSNE(n_components = 2, init = 'pca', random_state = 0)
X = tsne.fit_transform(std_df)
```

```python
fig, ax = plt.subplots()
sc = ax.scatter(X[:,0], X[:,1], c=out_df['labels'],cmap='viridis')
ax.legend(*sc.legend_elements(), title='clusters')
```

```python
from sklearn.metrics import davies_bouldin_score as dunn_index
```

```python
print(f'Dunn Index: {dunn_index(std_df, dbs.labels_)}')
```

```python
print(f'Silhouette Score: {silhouette_score(std_df, dbs.labels_)}')
```

```python
print(f'Time taken: {end - start}')
```
